```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
#import pdfkit as pdf
```

```python
prob_ = "signed_graphs"
n_ = 60
```

```python
form_ = "edge2"
method_ = "lp"

file=f"{method_}_n{n_}_{form_}_signed_default.txt"

df = pd.DataFrame()
df = pd.read_csv(file,header=None,sep=';')

lp2_n60 = pd.DataFrame()
lp2_n60 = pd.concat([lp2_n60, df], ignore_index=True)
lp2_n60.columns = ['instance','bs','time']
#lp2_n60

tab = pd.DataFrame({
    'instance':"lp2_default",
    'bs':lp2_n60['bs'].mean(),
    'time':lp2_n60['time'].mean()
    },index=[""]
)

lp2_n60 = pd.concat([lp2_n60,tab])
lp2_n60['instance'] = lp2_n60['instance'].str.replace('data/kmbs/RANDOM/random_','')
lp2_n60
```

```python
form_ = "edge2"
method_ = "mip"

file=f"{method_}_n{n_}_{form_}_signed_default.txt"

df = pd.DataFrame()
df = pd.read_csv(file,header=None,sep=';')

mip2_n60 = pd.DataFrame()
mip2_n60 = pd.concat([mip2_n60, df], ignore_index=True)
mip2_n60.columns = ['instance','bb','bs','gap','time','nodes','opt']
#mip2_n60

tab = pd.DataFrame({
    'instance':"mip2_default",
    'bb':mip2_n60['bb'].mean(),
    'bs':mip2_n60['bs'].mean(),
    'gap':mip2_n60['gap'].mean(),
    'time':mip2_n60['time'].mean(),
    'nodes':mip2_n60['nodes'].mean(),
    'opt':mip2_n60['opt'].sum()
    },index=[""]
)

mip2_n60 = pd.concat([mip2_n60,tab])
mip2_n60['instance'] = mip2_n60['instance'].str.replace('data/kmbs/RANDOM/random_','')
mip2_n60
```

```python
mip2_n60['instance'] = mip2_n60['instance'].str.replace('_','\_') 
print(mip2_n60[['instance','bs','gap','time','nodes','opt']].to_latex(index=False,float_format="%.2f"))
```
