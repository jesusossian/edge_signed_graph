```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
#import pdfkit as pdf
```

```python
prob_ = "signed_graphs"
n_ = 60
```

```python
form_ = "edge3"
method_ = "lp"

file=f"{method_}_n{n_}_{form_}_signed_default.txt"

df = pd.DataFrame()
df = pd.read_csv(file,header=None,sep=';')

lp3_n60 = pd.DataFrame()
lp3_n60 = pd.concat([lp3_n60, df], ignore_index=True)
lp3_n60.columns = ['instance','bs','time']
#lp3_n60

tab = pd.DataFrame({
    'instance':"lp3_default",
    'bs':lp3_n60['bs'].mean(),
    'time':lp3_n60['time'].mean()
    },index=[""]
)

lp3_n60 = pd.concat([lp3_n60,tab])
lp3_n60['instance'] = lp3_n60['instance'].str.replace('data/kmbs/RANDOM/random_','')
lp3_n60
```

```python
form_ = "edge3"
method_ = "mip"

file=f"{method_}_n{n_}_{form_}_signed_default.txt"

df = pd.DataFrame()
df = pd.read_csv(file,header=None,sep=';')

mip3_n60 = pd.DataFrame()
mip3_n60 = pd.concat([mip3_n60, df], ignore_index=True)
mip3_n60.columns = ['instance','bb','bs','gap','time','nodes','opt']
#mip3_n60

tab = pd.DataFrame({
    'instance':"mip3_default",
    'bb':mip3_n60['bb'].mean(),
    'bs':mip3_n60['bs'].mean(),
    'gap':mip3_n60['gap'].mean(),
    'time':mip3_n60['time'].mean(),
    'nodes':mip3_n60['nodes'].mean(),
    'opt':mip3_n60['opt'].sum()
    },index=[""]
)

mip3_n60 = pd.concat([mip3_n60,tab])
mip3_n60['instance'] = mip3_n60['instance'].str.replace('data/kmbs/RANDOM/random_','')
mip3_n60
```

```python
mip3_n60['instance'] = mip3_n60['instance'].str.replace('_','\_') 
print(mip3_n60[['instance','bs','gap','time','nodes','opt']].to_latex(index=False,float_format="%.2f"))
```
