```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
#import pdfkit as pdf
```

```python
prob_ = "signed_graphs"
n_ = 60
```

```python
form_ = "edge1"
method_ = "lp"

file=f"{method_}_n{n_}_{form_}_signed_default.txt"

df = pd.DataFrame()
df = pd.read_csv(file,header=None,sep=';')

lp1_n60 = pd.DataFrame()
lp1_n60 = pd.concat([lp1_n60, df], ignore_index=True)
lp1_n60.columns = ['instance','bs','time']
#lp1_n60

tab = pd.DataFrame({
    'instance':"resume",
    'bs':lp1_n60['bs'].mean(),
    'time':lp1_n60['time'].mean()
    },index=[""]
)

lp1_n60 = pd.concat([lp1_n60,tab])
lp1_n60['instance'] = lp1_n60['instance'].str.replace('data/kmbs/RANDOM/random_','')
lp1_n60
```

```python
form_ = "edge1"
method_ = "mip"

file=f"{method_}_n{n_}_{form_}_signed_default.txt"

df = pd.DataFrame()
df = pd.read_csv(file,header=None,sep=';')

mip1_n60 = pd.DataFrame()
mip1_n60 = pd.concat([mip1_n60, df], ignore_index=True)
mip1_n60.columns = ['instance','bb','bs','gap','time','nodes','opt']
#mip_n60

tab = pd.DataFrame({
    'instance':"resume",
    'bb':mip1_n60['bb'].mean(),
    'bs':mip1_n60['bs'].mean(),
    'gap':mip1_n60['gap'].mean(),
    'time':mip1_n60['time'].mean(),
    'nodes':mip1_n60['nodes'].mean(),
    'opt':mip1_n60['opt'].sum()
    },index=[""]
)

mip1_n60 = pd.concat([mip1_n60,tab])
mip1_n60['instance'] = mip1_n60['instance'].str.replace('data/kmbs/RANDOM/random_','')
mip1_n60
```
